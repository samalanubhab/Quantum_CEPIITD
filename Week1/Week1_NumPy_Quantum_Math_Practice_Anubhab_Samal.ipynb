## Question 1, Complex Numbers

Let

$$z = 5 - 2i$$

Using NumPy, compute and print the real part, the imaginary part, the modulus, and the conjugate of $z$.

Then verify the identity

$$|z|^2 = z z^*$$

by computing both sides and printing them.

## Question 2, Vectors and Kets

Define the two basic qubit states

$$|0\rangle = \begin{pmatrix} 1 \\ 0 \end{pmatrix}, \qquad |1\rangle = \begin{pmatrix} 0 \\ 1 \end{pmatrix}$$

as NumPy arrays.

Using these, construct the state

$$|-\rangle = \frac{1}{\sqrt{2}} \big(|0\rangle - |1\rangle\big)$$

Print the resulting vector, then compute and print its norm. The norm should be very close to one.

```python
# Your code here
ket_0=np.array([1,0])
ket_1=np.array([0,1])
ket_minus=1/np.sqrt(2)*(ket_0-ket_1)
print("ket_minus:",ket_minus)
norm=np.linalg.norm(ket_minus)
print("norm:",norm)
```

## Question 3, Inner Products

Let

$$u = \begin{pmatrix} 2 - i \\ 3 \end{pmatrix}, \qquad v = \begin{pmatrix} 1 \\ 2 + i \end{pmatrix}$$

Compute $\langle u, v \rangle$ using the correct NumPy function for a quantum mechanical inner product.

Then check whether $u$ and $v$ are orthogonal.

```python
# Your code here
u=np.array([2-1j,3])
v=np.array([1,2+1j])
inner_product=np.vdot(u,v)
print("inner_product:",inner_product)

```

## Question 4, Matrices

Let

$$A = \begin{pmatrix} 0 & 1 \\ 1 & 0 \end{pmatrix}, \qquad B = \begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix}$$

Compute and print the following four quantities.

1. $A + B$
2. The true matrix product $AB$
3. The Hermitian transpose of $A$
4. $A$ multiplied by the vector $|1\rangle = \begin{pmatrix} 0 \\ 1 \end{pmatrix}$

```python
# Your code here
A=np.array([[0,1],[1,0]])
B=np.array([[1,0],[0,-1]])
sum_AB=A+B
print("sum_AB:",sum_AB)
product_AB=A @ B
print("product_AB:",product_AB)
hermitian_transpose_A=np.conj(A.T)
print("hermitian_transpose_A:",hermitian_transpose_A)
ket_1=np.array([0,1])
product_A_ket_1=A @ ket_1
print("product_A_ket_1:",product_A_ket_1)

```
